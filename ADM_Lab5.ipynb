1. Load Dataset:**bold text**

```python
import pandas as pd
from google.colab import files
```

```python
data = pd.read_csv('titanic.csv')
```

2. Data Overview: **bold text**

```python
data.head()
```

```python
 data.info()
```

```python
data.describe()
```

```python
data.isnull().sum()
```

```python
data['Age'].fillna(data['Age'].median(), inplace=True)
data['Embarked'].fillna(data['Embarked'].mode()[0], inplace=True)
```

```python
print(data.isnull().sum())
```

```python
# Select only numerical features for correlation calculation
numerical_features = data.select_dtypes(include=['number'])
correlation_matrix = numerical_features.corr()

# Display correlation matrix
print(correlation_matrix)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f',
cbar=True)
plt.title('Correlation Matrix')
plt.show()
```

```python
 plt.figure(figsize=(8, 6))
sns.scatterplot(x='Age', y='Fare', data=data, hue='Survived',
palette='Set1')
plt.title('Age vs Fare: Trend Analysis')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.histplot(data['Age'], bins=30, kde=True, color='skyblue')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x=data['Fare'], color='lightcoral')
plt.title('Fare Outliers')
plt.xlabel('Fare')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.countplot(x='Pclass', data=data, palette='Set2')
plt.title('Passenger Class Distribution')
plt.xlabel('Pclass')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.countplot(x='Survived', data=data, palette='Set2')
plt.title('Survival Distribution')
plt.xlabel('Survived')
plt.ylabel('Count')
plt.show()
```
